```python
import chromadb

from pathlib import Path
from sentence_transformers import SentenceTransformer

PROJECT_ROOT = Path(
    r"C:\Users\sreej\Sales-Operations-Intelligence-Copilot"
)

CHROMA_DIR = PROJECT_ROOT / "data" / "vector_db"

# Connect to existing database
client = chromadb.PersistentClient(
    path=str(CHROMA_DIR)
)

# Access existing collection
collection = client.get_collection(
    name="company_knowledge"
)

# Load the same embedding model
embedding_model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)

# Test question
query = "What products and services does the company provide?"

# Generate query embedding
query_embedding = embedding_model.encode(
    query
).tolist()

# Retrieve relevant chunks
results = collection.query(
    query_embeddings=[query_embedding],
    n_results=3
)

print("QUESTION:", query)
print("\nRETRIEVED DOCUMENTS:\n")

for i, document in enumerate(results["documents"][0], 1):
    print(f"RESULT {i}")
    print(document)
    print("-" * 60)
```

```python
# Combine retrieved documents into context

retrieved_context = "\n\n".join(
    results["documents"][0]
)

question = "What products and services does the company provide?"

prompt = f"""
You are a company knowledge assistant.

Answer the question using ONLY the information
provided in the context.

If the answer is not available in the context,
clearly say that the information is unavailable.

CONTEXT:
{retrieved_context}

QUESTION:
{question}

ANSWER:
"""

print(prompt)
```
