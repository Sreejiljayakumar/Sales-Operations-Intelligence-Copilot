```python

import ollama
import logging

MODEL_NAME = "llama3.2:3b"

# Configure logging
logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

logger = logging.getLogger(__name__)


def check_ollama_connection():

    """Check whether Ollama is accessible and the model is available."""

    try:
        response = ollama.list()

        available_models = [
            model.model for model in response.models
        ]

        if MODEL_NAME not in available_models:
            logger.error(f"Model not found: {MODEL_NAME}")
            return False

        logger.info("Ollama connection successful.")
        logger.info(f"Model available: {MODEL_NAME}")

        return True

    except Exception as error:
        logger.exception("Ollama connection failed: %s", error)
        return False


def generate_response(prompt):

    """Generate an LLM response with basic error handling."""

    if not isinstance(prompt, str) or not prompt.strip():
        raise ValueError("Prompt cannot be empty.")

    try:

        response = ollama.chat(
            model=MODEL_NAME,
            messages=[
                {
                    "role": "user",
                    "content": prompt
                }
            ]
        )

        answer = response["message"]["content"]

        if not answer or not answer.strip():
            raise RuntimeError("Ollama returned an empty response.")

        return answer.strip()

    except ollama.ResponseError as error:
        logger.exception("Ollama model response error: %s", error)
        raise RuntimeError(
            "Ollama could not process the request. "
            "Check the model and prompt."
        ) from error

    except Exception as error:
        logger.exception("LLM generation failed: %s", error)
        raise RuntimeError(
            "Unable to generate an AI response. "
            "Check whether Ollama is running."
        ) from error


if __name__ == "__main__":

    print("\nChecking Ollama connection...")

    if check_ollama_connection():

        question = "What is a sales lead?"

        print("\nGenerating response...")

        try:
            answer = generate_response(question)

            print("\nLLM RESPONSE:")
            print(answer)

        except RuntimeError as error:
            print("ERROR:", error)

    else:
        print("Ollama is unavailable or the model is missing.")
```
